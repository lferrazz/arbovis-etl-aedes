```python
import pandas as pd
from pysus.api import PySUSClient

async def baixar_dengue_2025():
    client = PySUSClient()
    
    arquivos = await client.query(dataset='SINAN', group='DENG', year=2025)
    if not arquivos:
        raise RuntimeError("nda encontrado")
    
    arquivo_local = await client.download(arquivos[0])
    df = pd.read_parquet(arquivo_local.path)
    return df

try:
    df_bruto_2025 = await baixar_dengue_2025()
    print(f"total de registros {len(df_bruto_2025):,}")
except Exception as e:
    print(f"Erro: {e}")
```
